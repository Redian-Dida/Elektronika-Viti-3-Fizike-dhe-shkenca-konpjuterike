# 4. Analiza e Markdown

Gjate përdorimit të Markdown vura re se strukturimi i tekstit,shkruarja e formulave,krijimi i tabelave behet shume me i thjeshte dhe me i shpejtë krahasuar me programet tradicionale.
Të gjitha simbolet dhe shenjat qe perdora shfaqeshin sakte dhe funksionin nese ishin vendosur si duhej dhe nuk ishin te veshtira per tu kuptuar.
Provat dhe testimet qe bera me markdown me konfirmuan se ai eshte nje mjet shume praktik për dokumentim.

# 5. Diskutimi
Sjellja e sistemit: Me perdorimin e  Markdown mesova se shkrimi i  dokumentit  behet ne kohe shume me te shpejte, pasi nuk ka nevoje per menu te nderlikuara formatimi, por mjafton perdorimi i sintakses se thjeshte te tekstit.
Burimet e devijimit: Gjate shkrimit vura re se ndonje harrese e vogel e simboleve ose e hapesirave, si te titujt apo listat ose formulat, bente qe teksti te mos interpretohej sakte.
Permiresime konkrete: Per te ardhmen, perdorimi i shtesave dhe strukturave me te avancuara të Markdown do ta beje krijimin e dokumenteve teknike edhe me efikas.
